```python
import pandas as pd
import os

# Definir las rutas de las carpetas según tu estructura
carpeta_origen = 'sav'
carpeta_destino = 'csv'

# Asegurarse de que la carpeta de destino exista (por si acaso)
os.makedirs(carpeta_destino, exist_ok=True)

# Obtener una lista de todos los archivos .sav en la carpeta de origen
archivos_sav = [f for f in os.listdir(carpeta_origen) if f.endswith('.sav')]

# Iterar sobre cada archivo para leerlo y convertirlo
for archivo in archivos_sav:
    ruta_sav = os.path.join(carpeta_origen, archivo)
    
    # Crear el nuevo nombre del archivo cambiando la extensión a .csv
    nombre_csv = archivo.replace('.sav', '.csv')
    ruta_csv = os.path.join(carpeta_destino, nombre_csv)
    
    try:
        print(f"Procesando: {archivo}...")
        
        # Leer el archivo .sav
        df = pd.read_spss(ruta_sav)
        
        # Guardar el DataFrame como archivo .csv (sin incluir el índice)
        df.to_csv(ruta_csv, index=False, encoding='utf-8')
        
        print(f"¡Éxito! Guardado como: {nombre_csv}\n")
        
    except Exception as e:
        print(f"Error al procesar {archivo}: {e}\n")

print("Proceso de conversión finalizado.")
```
